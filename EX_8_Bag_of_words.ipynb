```python
from sklearn.feature_extraction.text import CountVectorizer

# Sample documents
documents = [
    "I enjoy machine learning",
    "Machine learning is interesting",
    "I enjoy learning new skills"
]
# Initialize CountVectorizer
vectorizer = CountVectorizer()
print(vectorizer)
```

```python
# Fit and transform the documents
X = vectorizer.fit_transform(documents)
print(X)
# Change it into dense matrix and display the answer.
print(vectorizer.vocabulary_)

# What does it represent??? Guess and share your answer.
```

```python
# Get the feature (word) names
feature_names = vectorizer.get_feature_names_out()

# Convert the result to array
bow_matrix = X.toarray()

# Display results
print("Vocabulary / Features:")
print(feature_names)

print("\nBag of Words Matrix:")
for i, row in enumerate(bow_matrix):
    print(f"Document {i+1}: {row}")
```
